```python
# ====================================================================
#   CYCLISTIC CASE STUDY _ Analisi de Datos Coursera
#   Autor: Mauricio Ramirez Cerda
#   Objetivo: Analizar diferencias entre miembros y usuarios casuales
# ===================================================================

# ============================
# 1. IMPORTAR LIBRERÍAS
# ============================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# ======================================
# Cargar Q2 (Datos de segundo trimestre)
# ======================================
df_q2 = pd.read_csv("Datos/Original/2019_Q2.csv")

#============================================================
# Renombrar columnas de Q2 para que coincidan con Q1, Q3 y Q4
#============================================================

df_q2 = df_q2.rename(columns={
    "01 - Rental Details Rental ID"                   : "trip_id",
    "01 - Rental Details Local Start Time"            : "start_time",
    "01 - Rental Details Local End Time"              : "end_time",
    "01 - Rental Details Bike ID"                     : "bikeid",
    "01 - Rental Details Duration In Seconds Uncapped": "tripduration",
    "03 - Rental Start Station ID"                    : "from_station_id",
    "03 - Rental Start Station Name"                  : "from_station_name",
    "02 - Rental End Station ID"                      : "to_station_id",
    "02 - Rental End Station Name"                    : "to_station_name",
    "User Type"                                       : "usertype",
    "Member Gender"                                   : "gender",
    "05 - Member Details Member Birthday Year"        : "birthyear"
})

#===============================
# Cargar Q1,Q3 y Q4
#==============================

df_q1 = pd.read_csv("Datos/Original/2019_Q1.csv")
df_q3 = pd.read_csv("Datos/Original/2019_Q3.csv")
df_q4 = pd.read_csv("Datos/Original/2019_Q4.csv")

# ============================
# Unir todo en un solo DF
# ============================

df = pd.concat([df_q1, df_q2, df_q3, df_q4], ignore_index=True)
print("Total de filas cargadas:", len(df))
```

```python
# ============================
# 3. LIMPIEZA Y UNIFICACIÓN
# ============================

# Convertir fechas
df['start_time'] = pd.to_datetime(df['start_time'])
df['end_time'] = pd.to_datetime(df['end_time'])

# Crear ride_length en segundos
df['ride_length'] = (df['end_time'] - df['start_time']).dt.total_seconds()

# Crear día de la semana
dias_es = {
    'Monday'   : 'Lunes',
    'Tuesday'  : 'Martes',
    'Wednesday': 'Miércoles',
    'Thursday' : 'Jueves',
    'Friday'   : 'Viernes',
    'Saturday' : 'Sábado',
    'Sunday'   : 'Domingo'
}
df['day_of_week'] = df['start_time'].dt.day_name()
df['day_of_week'] = df['day_of_week'].map(dias_es)

# Eliminar viajes con duración negativa o nula
df = df[df['ride_length'] > 0].copy()

# Estandarizar etiquetas de usuario
df['usertype'] = df['usertype'].replace({
    'Subscriber': 'member',
    'Customer': 'casual'
})

```

```python
# ============================
# 4. ANÁLISIS DESCRIPTIVO
# ============================

print("\nDuración promedio por tipo de usuario:")
print(df.groupby('usertype')['ride_length'].mean())

print("\nNúmero de viajes por día de la semana:")
print(df.groupby(['usertype', 'day_of_week'])['trip_id'].count())
```

```python
# ============================
# 5. VISUALIZACIONES
# ============================
plt.figure(figsize=(8,5))
sns.barplot(
    data=df,
    x='usertype',
    y='ride_length',
    estimator=np.mean,
    hue='usertype',
    palette='viridis',
    legend=False
)
plt.title("Duración promedio del viaje (segundos)")
plt.xlabel("Tipo de usuario")
plt.ylabel("Duración promedio (seg)")
plt.show()



```
